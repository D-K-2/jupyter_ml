```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import warnings
warnings.filterwarnings('ignore')
df=pd.read_csv("sonar_dataset.csv",header=None)
df.sample(5)
```

```python
df.shape
```

```python
df[60].value_counts()
```

```python
x=df.drop(60,axis=1)
```

```python
x
```

```python
y=df[60]
```

```python
y
y.shape
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
y=le.fit_transform(y)
```

```python
y
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model=keras.Sequential([
    keras.layers.Input(shape=(60,)),
    keras.layers.Dense(120,activation='relu'),
    keras.layers.Dense(60,activation="sigmoid"),
    keras.layers.Dense(20,activation="relu"),
    keras.layers.Dense(1,activation="sigmoid")])
```

```python
model.compile(loss="binary_crossentropy",optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=100,batch_size=15)
```

```python
model.evaluate(x_test,y_test)
```

```python
y_pred=model.predict(x_test).reshape(-1)
y_pred.shape
```

```python
y_pred=np.round(y_pred)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import confusion_matrix,classification_report
print(classification_report(y_test,y_pred))
```

```python
model_dropout_layer=keras.Sequential([
    keras.layers.Input(shape=(60,)),
    keras.layers.Dense(80,activation='relu'),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(30,activation="sigmoid"),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(20,activation="relu"),
    
    keras.layers.Dense(1,activation="sigmoid")
])
```

```python
model_dropout_layer.compile(loss="binary_crossentropy",optimizer='adam',metrics=['accuracy'])
```

```python
model_dropout_layer.fit(x_train,y_train,epochs=100,batch_size=15)
```

```python
model_dropout_layer.evaluate(x_test,y_test)
```

```python
y_pred=model_dropout_layer.predict(x_test).reshape(-1)
y_pred.shape
y_pred
y_pred=np.round(y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
